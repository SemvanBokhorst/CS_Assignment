```python
# packages
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# database
db = pd.read_csv('istherecorrelation.csv', sep=';', decimal = ',')
```

```python
plt.figure(dpi=300)

# Bar plot
plt.bar(db['Year'], db['NL Beer consumption [x1000 hectoliter]'])

plt.title('NL Beer Consumption Over Years', fontsize=14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Beer Consumption [1000 Hectoliters]')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

# Interpretation
The annual beer consumption in the Netherlands seems to be constant in the inital 10 years, but increases each year after that!
